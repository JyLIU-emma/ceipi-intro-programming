## Analogies for programming
1. A recipe for **pasta with Bolognese sauce**:

   - Ingredients: pasta, tomatoes, minced beef, onion, olive oil, salt, spices, etc.
   - Containers and utensils for preparing and serving the food: a pan, a plate, etc.

2. Steps:
   - Step 1: prepare the sauce:
     - Chop the tomatoes and onion.
     - Heat the pan.
     - Add olive oil, minced beef, tomatoes, spices, etc., and cook the sauce.
   - Step 2: cook the pasta (input: uncooked pasta; output: cooked pasta).
   - Step 3: mix the cooked pasta with the sauce in the pan, then serve it on a plate.

3. Interpreting the recipe using programming concepts:

   - **Variables**: we can choose different values for the ingredients.
     - Pasta: penne, fusilli, spaghetti, etc.
     - Sauce: Bolognese, pesto, etc.

   - **Functions**:
     - We could buy ready-made Bolognese sauce from a supermarket: the details of step 1 and its ingredients are packaged together, much like a function hides the details of a task.
     - `ingredients → cooking steps → pasta dish` corresponds to `input → function → output`.
     - In programming, inputs are data to process: text whose font should change, a search term, an image to crop, audio to play, a file to save, etc.

   - **Imports**: combine functions to do more.
     - Once a function exists, we can reuse it in another program by importing it, like picking up a ready-made tool.
     - In the physical world, duplicating a tool requires materials and time. For example, two axes require two pieces of iron and two handles. In the digital world, the additional cost of copying existing code is very small.
     - Some steps or ingredients in the recipe can be viewed as imported functions or outputs from earlier functions: cooking pasta, making pasta from flour and water, or mincing beef. The recipe already builds on earlier cooking processes.
     - A collection of functions can form software. Software can grow by adding functions or interact with other software to form an ecosystem, such as Google products or the Zotero extension for Chrome.

4. Another analogy: Lego

Imagine programming as building a Lego house:
- Basic pieces: bricks for walls, a tree, or a roof.
- Assembled pieces (functions): identical trees reused to make a garden, or a section combining walls and a roof.

### Conditions: `if`

#### Example 1: conditional discounts
Write a function that adjusts the total price as follows:
- Students (status `student`) pay 75% of the total price.
- Clients visiting during happy hour (status `happy`) pay 85% of the total price.
- Other clients (status `normal`) pay the full price.

```python
def price_after_discount(total_price):
  client_status = input("Is the client a student (student), visiting during happy hour (happy), or a regular client (normal)? ")
  if client_status == "student":
    print("The client is a student and pays 75% of the total price.")
    final_price = total_price * 0.75
  elif client_status == "happy":
    print("The client is visiting during happy hour and pays 85% of the total price.")
    final_price = total_price * 0.85
  else:
    print("The client must pay the regular price.")
    final_price = total_price
  return final_price
```

```python
price_after_discount(100)
# Only "student" and "happy" trigger a discount. Any other string leads to the regular price.
# These comparisons are case-sensitive.
```

#### Example 2: Right of withdrawal (simplified version)

Under French consumer law, a consumer generally has 14 days to withdraw from a distance contract. For the purchase of goods, this period normally begins when the consumer receives the product. However, the right of withdrawal does not generally apply to products made to the consumer’s specifications or clearly personalised.

For teaching purposes, this simplified simulator considers only three conditions: whether the contract was concluded at a distance, whether the product is personalised, and whether it was received no more than 14 days ago. It does not cover every legal exception and should not be treated as legal advice.

```python
from IPython.display import Image, display

display(Image(filename="data/withdrawal_right_flowchart.png", height=400))
```

```python
def check_withdrawal_right(
    is_distance_contract,
    is_personalised,
    days_since_delivery
):
    """Check whether the consumer has a withdrawal right."""
    if not is_distance_contract:
        conclusion = "No withdrawal right: the contract was not concluded at a distance."
    elif is_personalised:
        conclusion = "No withdrawal right: the product is personalised."
    elif days_since_delivery <= 14:
        conclusion = "The consumer has a withdrawal right."
    else:
        conclusion = "No withdrawal right: the 14-day period has expired."
    return conclusion



result = check_withdrawal_right(
    is_distance_contract=True,
    is_personalised=False,
    days_since_delivery=16
)

print(result)
```
