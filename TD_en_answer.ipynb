## Analogies for programming
1. A recipe for **pasta with Bolognese sauce**:

   - Ingredients: pasta, tomatoes, minced beef, onion, olive oil, salt, spices, etc.
   - Containers and utensils for preparing and serving the food: a pan, a plate, etc.

2. Steps:
   - Step 1: prepare the sauce:
     - Chop the tomatoes and onion.
     - Heat the pan.
     - Add olive oil, minced beef, tomatoes, spices, etc., and cook the sauce.
   - Step 2: cook the pasta (input: uncooked pasta; output: cooked pasta).
   - Step 3: mix the cooked pasta with the sauce in the pan, then serve it on a plate.

3. Interpreting the recipe using programming concepts:

   - **Variables**: we can choose different values for the ingredients.
     - Pasta: penne, fusilli, spaghetti, etc.
     - Sauce: Bolognese, pesto, etc.

   - **Functions**:
     - We could buy ready-made Bolognese sauce from a supermarket: the details of step 1 and its ingredients are packaged together, much like a function hides the details of a task.
     - `ingredients → cooking steps → pasta dish` corresponds to `input → function → output`.
     - In programming, inputs are data to process: text whose font should change, a search term, an image to crop, audio to play, a file to save, etc.

   - **Imports**: combine functions to do more.
     - Once a function exists, we can reuse it in another program by importing it, like picking up a ready-made tool.
     - In the physical world, duplicating a tool requires materials and time. For example, two axes require two pieces of iron and two handles. In the digital world, the additional cost of copying existing code is very small.
     - Some steps or ingredients in the recipe can be viewed as imported functions or outputs from earlier functions: cooking pasta, making pasta from flour and water, or mincing beef. The recipe already builds on earlier cooking processes.
     - A collection of functions can form software. Software can grow by adding functions or interact with other software to form an ecosystem, such as Google products or the Zotero extension for Chrome.

4. Another analogy: Lego

- Basic pieces: bricks for walls, a tree, or a roof.

#### Example 1: conditional discounts
Write a function that adjusts the total price as follows:
- Students (status `student`) pay 75% of the total price.
- Clients visiting during happy hour (status `happy`) pay 85% of the total price.
- Other clients (status `normal`) pay the full price.

````python
def price_after_discount(total_price):
  client_status = input("Is the client a student (student), visiting during happy hour (happy), or a regular client (normal)? ")
  if client_status == "student":
    print("The client is a student and pays 75% of the total price.")
    final_price = total_price * 0.75
  elif client_status == "happy":
    print("The client is visiting during happy hour and pays 85% of the total price.")
    final_price = total_price * 0.85
  else:
    print("The client must pay the regular price.")
    final_price = total_price
  return final_price
````

Under French consumer law, a consumer generally has 14 days to withdraw from a distance contract. For the purchase of goods, this period normally begins when the consumer receives the product. However, the right of withdrawal does not generally apply to products made to the consumer’s specifications or clearly personalised.

For teaching purposes, this simplified simulator considers only three conditions: whether the contract was concluded at a distance, whether the product is personalised, and whether it was received no more than 14 days ago.
````python
def check_withdrawal_right(
    is_distance_contract,
    is_personalised,
    days_since_delivery
):
    """Check whether the consumer has a withdrawal right."""
    if not is_distance_contract:
        conclusion = "No withdrawal right: the contract was not concluded at a distance."
    elif is_personalised:
        conclusion = "No withdrawal right: the product is personalised."
    elif days_since_delivery <= 14:
        conclusion = "The consumer has a withdrawal right."
    else:
        conclusion = "No withdrawal right: the 14-day period has expired."
    return conclusion



result = check_withdrawal_right(
    is_distance_contract=True,
    is_personalised=False,
    days_since_delivery=16
)

print(result)
````

Write a prompt asking an LLM to create a function that masks dates in text. For example:
- input: "On 10/02/1872, Phileas Fogg accepted a bet."
- output: "On [DATE], Phileas Fogg accepted a bet."

````python
import re


def mask_dates(text: str) -> str:
    # Matches patterns like DD/MM/YYYY, MM-DD-YYYY, YYYY-MM-DD, etc.
    date_pattern = r"\b(?:\d{1,4}[-/\.]\d{1,2}[-/\.]\d{1,4})\b"
    return re.sub(date_pattern, "[DATE]", text)


# Example usage:
sentence = "On 10/02/1872, Phileas Fogg accepted a bet."
print(mask_dates(sentence))
# Output: "On [DATE], Phileas Fogg accepted a bet."
````

Then, you encounter dates containing only a month and a day, or only a year, as well as dates written like this:
- "Phileas Fogg, having won twenty guineas at whist, and taken leave of his friends, left the Reform Club at ten minutes to eight. Passepartout had conscientiously studied his program; and it was noted down that on Wednesday, 2nd October, they were to start from London at 8:45 P.M."

Work with an LLM to develop a new function that can handle all these cases.

Possible approaches (you can include these technical details in your prompt):

- Pre-Trained NER(=Named Entity Recognition) models via spaCy (something closer to modern AI, machine learning)

````python
# code written by Gemini

# 1. use regex to find dates in various formats
# TODO:

# one possible Answer:
import re

MONTHS = (
    r"(?:Jan(?:uary)?|Feb(?:ruary)?|Mar(?:ch)?|Apr(?:il)?|May|Jun(?:e)?|"
    r"Jul(?:y)?|Aug(?:ust)?|Sep(?:tember)?|Oct(?:ober)?|Nov(?:ember)?|Dec(?:ember)?)"
)
DAYS_OF_WEEK = r"(?:Monday|Tuesday|Wednesday|Thursday|Friday|Saturday|Sunday)"

PATTERNS = [
    # 1. Date including the day of the week: "Wednesday, 2nd October 1872" or "Wednesday, 2nd October"
    re.compile(
        rf"\b{DAYS_OF_WEEK},\s+\d{{1,2}}(?:st|nd|rd|th)?\s+{MONTHS}(?:\s+\d{{4}})?\b",
        re.IGNORECASE,
    ),
    # 2. Standard ISO / Numeric: "1872-10-02", "10/02/1872", "02.10.1872"
    re.compile(
        r"\b(?:\d{4}[-/.]\d{1,2}[-/.]\d{1,2}|\d{1,2}[-/.]\d{1,2}[-/.]\d{2,4})\b"
    ),
    # 3. Month & Day (both orders): "2nd October", "October 2nd", "Oct 2, 1872"
    re.compile(
        rf"\b(?:\d{{1,2}}(?:st|nd|rd|th)?\s+(?:of\s+)?{MONTHS}|{MONTHS}\s+\d{{1,2}}(?:st|nd|rd|th)?)(?:[,\s]+\d{{4}})?\b",
        re.IGNORECASE,
    ),
    # 4. Standalone Month + Year: "October 1872"
    re.compile(rf"\b{MONTHS}\s+\d{{4}}\b", re.IGNORECASE),
    # 5. Standalone Year with contextual preposition: "in 1872", "during 1905"
    re.compile(r"\b(?:in|during|by|around|year)\s+(\d{4})\b", re.IGNORECASE),
]


def mask_dates_regex(text: str, mask_token: str = "[DATE]") -> str:
    spans = []

    # 1. get date spans
    for pattern in PATTERNS:
        for match in pattern.finditer(text):
            start, end = match.span()
            # Ensure this match doesn't overlap with an already captured longer date
            if not any(
                s_start <= start and end <= s_end for s_start, s_end in spans
            ):
                spans.append((start, end))

    # 2. sort spans by their position in the text, from back to front
    spans.sort(key=lambda x: x[0], reverse=True)

    # 3. replace spans with mask token
    masked_text = text
    for start, end in spans:
        masked_text = masked_text[:start] + mask_token + masked_text[end:]

    return masked_text
````

````python
# 2. use pre-trained models via spaCy

# One possible answer:
import spacy

# Load the English model outside of the function to avoid reloading it every time the function is called.
nlp = spacy.load("en_core_web_sm")


def mask_dates_spacy(
    text: str, mask_token: str = "[DATE]", model=nlp
) -> str:
    """Use spaCy to locate all entities labeled as DATE and replace them with the mask token."""
    doc = model(text)

    # 1. Filter out all entities labeled as DATE
    date_ents = [ent for ent in doc.ents if ent.label_ == "DATE"]

    # 2. Sort by character start position from back to front (reverse order replacement to prevent index offset)
    date_ents.sort(key=lambda ent: ent.start_char, reverse=True)

    # 3. Replace spans with mask token
    masked_text = text
    for ent in date_ents:
        masked_text = (
            masked_text[: ent.start_char] + mask_token + masked_text[ent.end_char :]
        )

    return masked_text
````

### 3. Validating LLM-generated code

Prompt used for creating this validation test:
```md
I want to create a test dataset and supporting code to evaluate a function that masks dates in text. The requirements are:

1. Create 100 English sentences, each containing 1–2 date expressions. Use a wide variety of formats, including ISO dates, standalone years, month-and-day combinations, and dates containing days of the week. Include both full names and abbreviations.

2. Save the dataset as a CSV file with three columns: `text`, `masked_text`, and `date_span`. For example:

- `text`: On Wednesday, 2nd October, Phileas Fogg accepted the bet.
- `masked_text`: On [DATE], Phileas Fogg accepted the bet.
- `date_span`: Wednesday, 2nd October

3. The function to test, `mask_dates_func`, already exists. It takes the original sentence from `text` as input and should return the expected result in `masked_text`.

Write code that tests the function on every sentence by comparing its output with the expected `masked_text`.

- If all tests pass, print `"function passed"` at the end.
- Otherwise, print the accuracy at the end.
- For each failed test, print the original sentence, the expected output, the actual output, and the date text that should have been identified.
```

````python
import csv

def check_mask_dates(mask_dates_func, csv_path="data/date_masking_test.csv"):
    """Check a masking function against the expected output in each CSV row."""
    correct = 0
    total = 0

    with open(csv_path, encoding="utf-8-sig", newline="") as file:
        for row in csv.DictReader(file):
            total += 1
            try:
                actual = mask_dates_func(row["text"])
                passed = actual == row["masked_text"]
            except Exception as error:
                actual = f"{type(error).__name__}: {error}"
                passed = False

            if passed:
                correct += 1
            else:
                print(f"\nFailed case {total}")
                print("Original:", row["text"])
                print("Expected:", row["masked_text"])
                print("Actual:", repr(actual))
                print("Expected date spans:", row["date_span"])

    if total == 0:
        print("No test cases found.")
    elif correct == total:
        print("function passed")
    else:
        print(f"Accuracy: {correct / total:.2%} ({correct}/{total})")
````

````python
# You only need to import the same module once in a Python script.
# import re

MONTHS = (
    r"(?:Jan(?:uary)?|Feb(?:ruary)?|Mar(?:ch)?|Apr(?:il)?|May|Jun(?:e)?|"
    r"Jul(?:y)?|Aug(?:ust)?|Sep(?:tember)?|Oct(?:ober)?|Nov(?:ember)?|Dec(?:ember)?)"
)

DAYS_OF_WEEK = (
    r"(?:Monday|Tuesday|Wednesday|Thursday|Friday|Saturday|Sunday)"
)

PATTERNS = [
    # 1. Date with weekday: "Wednesday, 2nd October 1872"
    re.compile(
        rf"\b{DAYS_OF_WEEK},\s+\d{{1,2}}(?:st|nd|rd|th)?"
        rf"\s+{MONTHS}(?:\s+\d{{4}})?\b",
        re.IGNORECASE,
    ),

    # 2. Numeric date: "1872-10-02", "10/02/1872", "02.10.1872"
    re.compile(
        r"\b(?:\d{4}[-/.]\d{1,2}[-/.]\d{1,2}|"
        r"\d{1,2}[-/.]\d{1,2}[-/.]\d{2,4})\b"
    ),

    # 3. Month and day: "2nd October", "October 2nd", "Oct 2, 1872"
    re.compile(
        rf"\b(?:"
        rf"\d{{1,2}}(?:st|nd|rd|th)?\s+(?:of\s+)?{MONTHS}|"
        rf"{MONTHS}\s+\d{{1,2}}(?:st|nd|rd|th)?"
        rf")(?:[,\s]+\d{{4}})?\b",
        re.IGNORECASE,
    ),

    # 4. Month and year: "October 1872"
    re.compile(
        rf"\b{MONTHS}\s+\d{{4}}\b",
        re.IGNORECASE,
    ),

    # 5. Year with context: capture the year without the preceding word.
    re.compile(
        r"\b(?:in|during|by|around|year)\s+(?P<date>\d{4})\b",
        re.IGNORECASE,
    ),
]


def mask_dates_regex_new(text: str, mask_token: str = "[DATE]") -> str:
    spans = []

    # Collect date positions in the original text.
    for pattern in PATTERNS:
        for match in pattern.finditer(text):

            # Use the named date group when the pattern provides one.
            if "date" in pattern.groupindex:
                start, end = match.span("date")
            else:
                start, end = match.span()

            # Skip matches that overlap with an accepted date.
            if not any(
                start < saved_end and end > saved_start
                for saved_start, saved_end in spans
            ):
                spans.append((start, end))

    # Replace from right to left to preserve earlier positions.
    spans.sort(reverse=True)

    for start, end in spans:
        text = text[:start] + mask_token + text[end:]

    return text
````
